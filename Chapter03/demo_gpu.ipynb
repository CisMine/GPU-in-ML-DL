```python
import time
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

start_time = time.time()


data = pd.read_csv('linkedin_job_postings.csv')

data = data.dropna()

encoder = LabelEncoder()


for col in data:
    data[col] = encoder.fit_transform(data[col])

y = data["job_level"]
x = data.drop("job_level", axis=1)

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=42)

model = RandomForestClassifier()
model.fit(x_train, y_train)

test_score = model.score(x_test, y_test)

print(f"Test score: {test_score}")



end_time = time.time()
execution_time = end_time - start_time
print(f"Execution time for CPU: {execution_time} seconds")

```

```python
import time
from cuml.preprocessing import LabelEncoder
from cuml.model_selection import train_test_split
from cuml.ensemble import RandomForestClassifier
import cudf

start_time = time.time()


data = cudf.read_csv('linkedin_job_postings.csv')

data = data.dropna()

encoder = LabelEncoder()


for col in data:
    data[col] = encoder.fit_transform(data[col])

y = data["job_level"]
x = data.drop("job_level", axis=1)

x_train, x_test, y_train, y_test = train_test_split(x, y, train_size=0.8, random_state=42)

model = RandomForestClassifier()
model.fit(x_train, y_train)

test_score = model.score(x_test, y_test)

print(f"Test score: {test_score}")



end_time = time.time()
execution_time = end_time - start_time
print(f"Execution time for GPU: {execution_time} seconds")

```
